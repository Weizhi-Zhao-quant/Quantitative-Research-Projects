# The Alligator Indicator

## What is the Alligator?

Bill Williams was an American Commodity Trading Advisor (CTA) with extensive experience in commodities, stocks, and foreign exchange trading. In his 1995 book Trading Chaos, he introduced the concept of the "Alligator" indicator, which reportedly achieved substantial returns in the US stock, forex, and futures markets at that time. Three decades later, we explore whether the Alligator strategy still holds its power in today's A-share market. Following the indicator construction and trading methodology from Dr. Williams' book, we will recreate the Alligator strategy. Given that technical analysis strategies like the Alligator involve numerous settings and parameters, to avoid overfitting in backtesting, we will largely adhere to the original parameters from the book when applying this strategy to the A-share market.

First, what is the "Alligator"? According to Trading Chaos, broadly speaking, the Alligator is a set of balance lines constructed based on fractal geometry and nonlinear dynamics.
The ​blue line​ is called the Alligator's ​Jaw. It is the moving average representing the chart's primary time frame. It represents the market's state in the absence of new information. The divergence between the blue line and the actual price action reflects traders' interpretation and reaction to new incoming information.
The ​red line​ is the moving average of a time frame one level lower than the chart's primary frame – specifically, about one-fifth of the primary cycle. For a daily chart, the blue line is a daily moving average, while the red line corresponds roughly to an hourly timeframe.
The ​green line​ is the moving average of an even lower time frame, approximately one-fifth of the red line's span, roughly equivalent to a 5-10 minute timeframe.

New information first affects the green line, then the red line, and finally causes changes in the blue line. Significant unexpected events impact traders, causing price action to deviate from the blue line. Evidently, the green line is most sensitive to incoming information, while the blue line is the most stable. For easier understanding, we metaphorically call the blue line the "Alligator's Jaw," the red line the "Alligator's Teeth," and the green line the "Alligator's Lips." As traders become familiar with these terms, the Alligator concept becomes more vivid. The Alligator indicator signals when to enter a trade or when to wait on the sidelines until the Alligator is "hungry" and opens its mouth. In summary, the Alligator provides:

1. An integrated method to observe market momentum across three different timeframes simultaneously on a single chart.

2. A simple way to determine when a trend starts and stops.

3. A protective mechanism when the market is consolidating without a clear direction.

4. An advanced indicator signaling a potential trend end.


![](img/01912535-9dd2-70ac-a7fd-da75d8beb2de_5_605520.jpg)

Frequently, markets enter consolidation phases where the three lines intertwine. This price action suggests traders should ideally remain on the sidelines, as there isn't significant new information to drive a trend. Markets often reach a temporary balance between buyers and sellers within a narrow range, which is also why trends often pause near significant support and resistance levels.

Major events or new information are required to draw sidelined participants into the market. This information is best when it is new or unexpected, prompting more traders to act, thereby increasing the volume of the traded stock or futures contract. Ultimately, driven by demand and supply, a significant trend emerges.

**​Bullish Scenario (Hungry Alligator)​:** With the arrival of positive news, a strong upward trend emerges. The three lines begin to diverge upwards. The Alligator awakens from sleep and starts moving up to "eat" the prices.

**​Bearish Scenario (Full Alligator)​:**Conversely, if negative news arrives, a downtrend begins. The Alligator opens its mouth (green, red, blue lines diverge downwards) and starts moving down to "eat" the prices.

**​Consolidation (Sleeping Alligator)​:** When the three Alligator lines are intertwined, it signifies the Alligator is in a prolonged sleep. In this case, traders should avoid entering and patiently wait for the Alligator to awaken. Trading during the Alligator's sleep phase often leads to losses due to market noise and whipsaws.


## Construction of the Alligator Indicator and Timing Strategy


The Alligator indicator consists of three moving averages of different periods: a long, medium, and short moving average. Their construction is as follows:
1. The moving average of 13 price bars, shifted 8 bars into the future (to the right).
2. The moving average of 8 price bars, shifted 5 bars into the future.
3. The moving average of 5 price bars, shifted 3 bars into the future.

The parameters for the moving average periods and forward shifts are taken directly from Williams' Trading Chaosto prevent parameter overfitting. Interestingly, these parameters (13, 8, 5, 3) are numbers from the Fibonacci sequence.

Next, we construct a timing signal based on the Alligator indicator for a broad-based index (using the CSI 300 as an example):

**​Bullish Signal (Hungry Alligator)​:** If on day T, the Alligator lines are in a bullish alignment (Lips > Teeth > Jaw), the signal is bullish, and the position is set to fully invested (100% long).

**​Bearish Signal (Full Alligator)​:** If on day T, the Alligator lines are in a bearish alignment (Lips < Teeth < Jaw), the signal is bearish, and the position is set to fully short or empty (0% long / cash).

**​No Signal (Sleeping Alligator)​:** For any other alignment, there is no new signal, and the position is maintained from the previous trading day.

The execution price is assumed to be the opening price of the index on day T+1. Transaction costs are temporarily not considered.

# Awesome Oscillator (AO)
In most cases, there is no clear trend in the market; only rarely does an easily identifiable trend emerge. Only floor traders and a few specialist traders can navigate such conditions, as most traders set their stop-losses too tight and are often shaken out by the volatile, range-bound price action. Significant profits are only possible when the market exhibits a sustained trend. Therefore, a trader's primary consideration is to avoid missing any major market moves. The Alligator indicator provides signals for clear uptrends or downtrends. However, due to the high speculative sentiment in the A-share market, "false breakouts" frequently occur in both indices and individual stocks. Thus, the selection of entry points is crucial. Next, we introduce other trend and breakout indicators to further improve the win rate and profit-to-loss ratio of the Alligator strategy.

Here, we introduce the Awesome Oscillator (AO). Williams wrote in his book: "Without a doubt, the AO is the best oscillator for trading commodities and stocks. With the AO, a trader can see the market's momentum as clearly as if reading tomorrow's Wall Street Journal today." The AO indicator is the simple moving average of 5 price bars minus the simple moving average of 34 price bars. The construction of the AO is simple yet effective; a trader can potentially profit in futures and stock markets using onlythe AO indicator.

## Construction Method of the Awesome Oscillator

The Awesome Oscillator provides traders with the key to profitability.

The AO measures the momentum change of the most recent 5 price bars relative to the most recent 34 price bars.

1.First, calculate the simple moving average of the midpoint between the high and low of the last 5 price bars, i.e., SMA((High - Low) / 2, 5).

2.Subtract from this value the simple moving average of the midpoint between the high and low of the last 34 price bars, i.e., SMA((High - Low) / 2, 34).

3.The difference between these two values is the AO indicator, which is displayed as a histogram.


The Awesome Oscillator shows the current change in momentum.

- AO = SMA(High + Low, 5) - SMA(High + Low, 34)

Price is the last element in the market to change. Before price changes, momentum changes; before momentum changes, the speed of the current momentum changes; before the speed of momentum changes, volume changes; and before volume changes, the perceptions and subsequent actions of all traders and investors change.

The Awesome Oscillator is the first entry signal in the early stage of a trend. Assuming the Alligator shows a bullish signal, if the AO indicator (which shows changes in market momentum) increases for three consecutive trading days, it serves as a good signal to add to the position. Therefore, the timing strategy based on the Alligator and the AO indicator is constructed as follows:

**​Bullish Signal:** If on day T, the Alligator lines are in a bullish alignment ​AND​ the AO indicator has been increasing for 3 consecutive days, the signal is bullish, and the position is fully invested.

**​Bearish Signal:** If on day T, the Alligator lines are in a bearish alignment ​OR​ the AO indicator has been decreasing for 3 consecutive days, the signal is bearish, and the position is empty (cash).

**​No Signal:** In all other cases, there is no signal, and the position is maintained from the previous trading day.

Note: The trigger condition for the bullish signal is stricter (requiring "AND" – both conditions must be met), while the trigger condition for the bearish signal is more lenient (requiring "OR" – only one condition needs to be met). This is because the A-share market involves high speculation with many "false breakouts," and downturns often occur rapidly. Therefore, entering long positions should be more cautious, while reacting to bearish signals should be more timely.

# Fractal Breakout Signal

## What is a Fractal?

A Fractal, originally a term in geometry proposed by Polish mathematician Benoit Mandelbrot, refers to a shape that fills space with a non-integer dimensional form. It is typically defined as a rough or fragmented geometric shape that can be split into parts, each of which is a reduced-scale copy of the whole, exhibiting self-similarity. In mathematics and geometry, the Koch snowflake, Sierpinski triangle, and Cantor set are all significant milestones in the development of fractal theory. In nature and the real world, snowflakes, Romanesco broccoli, crystals, and even the coastline of Britain are closely related to fractal structures.

In financial markets, the fractal pattern is not overly complex. The market always moves in one direction or another. After a period, all bullish traders have entered the market (in an uptrend). Due to exhausted buying power or short-term sellers entering, the market begins to pull back. Subsequently, new information influences traders' judgments, new buying emerges, and the market starts rising again. If the market's momentum or bullish power at that time is sufficient for the price to break through the previous upward fractal, traders should buy above the high point of that fractal.

Pattern A in the figure below is a standard fractal pattern (upward fractal). The definition of an upward fractal is: it consists of at least 5 consecutive price bars. The highest high has two lower highs to its left and right. To the left of the left high and to the right of the right high, there are also two lower highs (the construction of a downward fractal is the opposite). A more visual method is to raise one hand in front of you, fingers spread with the middle finger pointing up. Your fingers represent the 5 consecutive price bars, and your middle finger is the highest high—this forms a fractal pattern.

![](img/01912535-9dd2-70ac-a7fd-da75d8beb2de_12_891808.jpg)

Pattern A is the original fractal pattern. Pattern B satisfies the definitions of both an upward and a downward fractal because the high of the middle fractal (middle finger) is higher than the highs of the two adjacent bars on both sides, while its low is also lower than the lows of those two adjacent bars. Pattern C shows upward and downward fractals appearing simultaneously within 6 price bars, sharing common bars. Pattern D is a fractal pattern formed by 6 price bars because the high of the 5th bar equals the previous highest high. In a buy fractal (upward fractal), traders only need to observe the high of each price bar; in a sell fractal (downward fractal), traders only need to observe the low of each price bar.

## Construction of the Fractal-Based Timing Strategy

The timing signal based on fractals is constructed as follows:

-**​Bullish Signal:** If on day T, the index closing price breaks upward through the highest price of the most recent upward fractal (the high of the 3rd bar in the fractal pattern), it is considered a breakout above resistance. The signal is bullish, and the position is fully invested.

-**​Bearish Signal:** If on day T, the index closing price breaks downward through the lowest price of the most recent downward fractal (the low of the 3rd bar in the fractal pattern), it is considered a breakdown below support. The signal is bearish, and the position is empty (cash).

-**​No Signal:** In all other cases, there is no signal, and the position is maintained from the previous trading day.

Next, we combine the Alligator strategy with the AO and Fractal signals, denoted as Alligator, AO, and Fractal respectively.

-​**Bullish Signal:​** If on day T, (Alligator == 1) AND (AO == 1 OR Fractal == 1), then the signal is bullish, and the position is fully invested.

-​**​Bearish Signal:​** If on day T, (Alligator == -1) OR (AO == -1) OR (Fractal == -1), then the signal is bearish, and the position is empty (cash).

-​**​No Signal:​** In all other cases, there is no signal, and the position is maintained from the previous trading day.

# MACD

MACD (Moving Average Convergence Divergence) is one of the most common technical analysis indicators, composed of the slow line (DEA), the fast line (DIF), the histogram, and the zero line. Specifically, DIF = 12-day EMA - 26-day EMA, DEA = 9-day EMA of DIF, and Histogram = (DIF - DEA) × 2. Note that the exponential smoothing periods for the fast line, slow line, and DEA use the default values of the MACD model, namely 12, 26, and 9, to avoid overfitting.

![](img/01912535-9dd2-70ac-a7fd-da75d8beb2de_15_737719.jpg)

If the DIF (fast line) crosses above the DEA (slow line), and simultaneously the histogram turns from green to red, it indicates that the price is starting to stop falling and rebound, suggesting a potential future rise; thus, the signal is bullish. Depending on the position of the bullish signal (above the zero line, below the zero line, or near the zero line), it can be categorized as:
•   Bullish Signal Above Water: Occurs above the zero line.

•   Bullish Signal Below Water: Occurs below the zero line.

•   Bullish Signal at Zero Line: Occurs near the zero line.

Although all are bullish signals implying a market recovery trend, their implications differ based on their location:
•   A bullish signal above water indicates the market is already in an uptrend, often signaling the start of a new significant rally.

•   A bullish signal below water suggests the market is in a downtrend. In this case, the bullish signal might only represent a brief rebound or a bull trap set by sellers, and the downtrend may resume afterward. Therefore, the reliability of bullish signals below water is relatively lower and typically requires multiple confirmations to prove a genuine market rebound.

•   A bullish signal at the zero line indicates the market is consolidating (or bottoming), with balanced forces between buyers and sellers. Once this balance is broken, a significant trend may emerge.

The logic applies similarly to bearish signals: bearish signals occurring during a downtrend (below water) or during consolidation (at the zero line) are generally more effective.

![](img/01912535-9dd2-70ac-a7fd-da75d8beb2de_14_976431.jpg)

Therefore, we construct the timing signal based on the following:
•   Bullish MACD Signal: If on day T, a bullish signal above water OR at the zero line is triggered.

•   Bearish MACD Signal: If on day T, a bearish signal below water OR at the zero line is triggered.

•   No Signal: In all other cases.

Next, we combine the Alligator strategy with the AO, Fractal, and MACD signals, denoted as Alligator, AO, Fractal, and MACD respectively.
•   Bullish Combined Signal: If on day T, (Alligator == 1) AND (Fractal == 1 OR MACD == 1), then the signal is bullish, and the position is fully invested.

•   Bearish Combined Signal: If on day T, (Alligator == -1) OR (AO == -1) OR (Fractal == -1) OR (MACD == -1), then the signal is bearish, and the position is empty (cash).

•   No Signal: In all other cases, there is no signal, and the position is maintained from the previous trading day.
